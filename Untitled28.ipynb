```python
import os
import re
import fitz  # PyMuPDF
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document # Corrected import path

# --- AYARLAR (Colab için güncellendi) ---
PDF_PATH = "tck.pdf"  # Colab'a yüklediğin PDF'in adı
FAISS_DB_PATH = "faiss_db"
# Model seçimi: Senin scriptindeki distiluse yerine daha güçlü olan E5-Large'ı kullanmanı öneririm
# Ama istersen seninkini de bırakabiliriz.
EMBEDDING_MODEL = "intfloat/multilingual-e5-large"

def extract_pdf_text(pdf_path):
    try:
        doc = fitz.open(pdf_path)
        text = ""
        for page in doc:
            text += page.get_text()
        return text
    except Exception as e:
        print(f"❌ Hata: PDF okunamadı: {e}")
        return None

def parse_tck_logic(text, dosya_adi):
    chunklar = []
    # Madde X- kalıbına göre böl
    madde_etiketleri = re.findall(r'Madde \d+-\s*', text)
    madde_icerikleri = re.split(r'Madde \d+-\s*', text)

    print(f"🔍 Toplam {len(madde_etiketleri)} madde tespit edildi.")

    for i, blok_icerigi in enumerate(madde_icerikleri[1:]):
        try:
            madde_etiketi = madde_etiketleri[i]
            madde_no = re.search(r'(\d+)', madde_etiketi).group(1)

            onceki_blok = madde_icerikleri[i]
            baslik_satirlari = [satir.strip() for satir in onceki_blok.strip().split('\n') if satir.strip()]
            madde_basligi = baslik_satirlari[-1] if baslik_satirlari else "Başlık Bulunamadı"

            mevcut_blok_satirlari = blok_icerigi.strip().split('\n')
            if (i + 1) < len(madde_etiketleri):
                temiz_metin = "\n".join(mevcut_blok_satirlari[:-1]).strip()
            else:
                temiz_metin = "\n".join(mevcut_blok_satirlari).strip()

            fikra_bolme_kalibi = r'(\(\d+\))'
            fikralar = re.split(fikra_bolme_kalibi, temiz_metin)

            j = 1
            while j < len(fikralar):
                fikra_no_etiketi = fikralar[j]
                fikra_icerigi = fikralar[j+1].strip() if (j+1) < len(fikralar) else ""

                # Zengin metin oluşturma
                zengin_metin = f"MADDE {madde_no} - {madde_basligi}\n{fikra_no_etiketi} {fikra_icerigi}"

                chunklar.append(Document(
                    page_content=zengin_metin,
                    metadata={
                        "source": dosya_adi,
                        "madde_no": madde_no,
                        "baslik": madde_basligi,
                        "fikra": fikra_no_etiketi
                    }
                ))
                j += 2
        except Exception as e:
            continue
    return chunklar

# --- ANA ÇALIŞTIRICI ---
print("🚀 İşlem Başlıyor...")
raw_text = extract_pdf_text(PDF_PATH)
if raw_text:
    documents = parse_tck_logic(raw_text, "tck.pdf")
    print(f"✅ Toplam {len(documents)} adet fıkra (chunk) oluşturuldu.")

    print("🧠 Vektör veritabanı oluşturuluyor...")
    # Colab'da GPU kullanmak için device:cuda ekledik
    embeddings = HuggingFaceEmbeddings(
        model_name=EMBEDDING_MODEL,
        model_kwargs={'device': 'cuda'}
    )

    vector_store = FAISS.from_documents(documents, embeddings)
    vector_store.save_local(FAISS_DB_PATH)
    print(f"🎉 BİTTİ! Veritabanı '{FAISS_DB_PATH}' klasörüne kaydedildi.")
```

```python
from unsloth import FastLanguageModel
import torch

# 1. Taban modeli doğrudan Meta'nın orijinal 4-bit versiyonundan çek (RAM dostu)
base_model_id = "unsloth/llama-3.1-8b-instruct-bnb-4bit"

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = base_model_id,
    max_seq_length = 2048,
    load_in_4bit = True,
    # RAM'i korumak için eklendi, cihaz eşleşmesini otomatikleştirir
    device_map = "auto",
    low_cpu_mem_usage = True,
)

# 2. Üzerine sadece SENİN hukuk uzmanlığı katmanını (adapter) giydir
model.load_adapter("Cetin003/hukuk_model_tck_v1_lora")

# Modelin tamamen CUDA üzerinde olduğundan emin ol
model.to('cuda')

FastLanguageModel.for_inference(model)
print("✅ RAM'i üzmeden sistemi ayağı kaldırdık!")
```

```python
import torch
from transformers import LlamaForCausalLM

# 1. Unsloth'un yamalarını çalışma zamanında (runtime) devre dışı bırakıyoruz
# Bu satırlar modelin içindeki 'akıllı' ama hatalı Unsloth fonksiyonlarını siler
if hasattr(model, "_old_generate"):
    model.generate = model._old_generate

# Unsloth'un 'fast forward' mekanizmasını tamamen bypass ediyoruz
model.forward = LlamaForCausalLM.forward.__get__(model, LlamaForCausalLM)

# 2. KV Cache ve KV Mask hatalarını önlemek için cache'i kapatıp tekrar açıyoruz
model.config.use_cache = False
# RAG'da değişken uzunlukta sorun çıkaran KV Cache ayarı
torch.cuda.empty_cache()

print("🔥 Unsloth'un tüm özel forward yolları imha edildi. Saf Transformers moduna geçildi.")

def hukuk_asistani_sor(soru):
    # k=2 kalsın, hem ana maddeyi hem nitelikli hali alsın
    docs = vector_store.similarity_search(f"query: {soru}", k=2)
    baglam = "\n\n".join([d.page_content for d in docs])

    # Talimatları daha sert ve net veriyoruz (System Prompt etkisi)
    prompt = f"""### Talimat:
Aşağıdaki TCK maddelerini bir hukuk uzmanı titizliğiyle incele.
Soruya cevap verirken mutlaka:
1. İlgili Madde numarasını belirt.
2. Varsa ceza artırımı gerektiren halleri (gece, silah vb.) açıkla.
3. Profesyonel ve ciddi bir dil kullan.

### İlgili Kanun Metinleri:
{baglam}

### Kullanıcı Sorusu:
{soru}

### Hukuki Değerlendirme:
"""

    inputs = tokenizer([prompt], return_tensors = "pt").to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            input_ids = inputs.input_ids,
            attention_mask = inputs.attention_mask,
            max_new_tokens = 450,
            temperature = 0.3, # Kararlılık için biraz düşürdük
            do_sample = True,
            use_cache = False,
            repetition_penalty = 1.2 # Aynı kelimeleri tekrarlamasını önler
        )

    res = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return res.split("### Hukuki Değerlendirme:")[-1].strip()

# --- SİSTEMİ ZORLAYAN BİR TEST ---
print("-" * 30)
soru = "Gece vakti silahla konut dokunulmazlığını ihlal eden birine verilecek ceza, temel cezaya göre nasıl değişir?"
print(hukuk_asistani_sor(soru))
```

```python
import torch
from transformers import LlamaForCausalLM

# 1. Unsloth'un yamalarını çalışma zamanında (runtime) devre dışı bırakıyoruz
# Bu satırlar modelin içindeki 'akıllı' ama hatalı Unsloth fonksiyonlarını siler
if hasattr(model, "_old_generate"):
    model.generate = model._old_generate

# Unsloth'un 'fast forward' mekanizmasını tamamen bypass ediyoruz
model.forward = LlamaForCausalLM.forward.__get__(model, LlamaForCausalLM)

# 2. KV Cache ve KV Mask hatalarını önlemek için cache'i kapatıp tekrar açıyoruz
model.config.use_cache = False
# RAG'da değişken uzunlukta sorun çıkaran KV Cache ayarı
torch.cuda.empty_cache()

def hukuk_asistani_sor(soru):
    # k=2 yapalım ki hem kasten hem taksirle gibi yakın maddeleri görsün
    docs = vector_store.similarity_search(f"query: {soru}", k=2)

    baglam = ""
    for i, d in enumerate(docs, 1):
        baglam += f"--- KANUN KAYNAĞI {i} ---\n{d.page_content}\n\n"

    prompt = f"""### TALİMAT:
Sen bir hukuk uzmanısın. Aşağıdaki KANUN KAYNAKLARI'nı incele ve SORU'ya en uygun olan maddeyi seçerek cevap ver.
Sadece kaynaklarda yazan bilgileri kullan, dışarıdan bilgi ekleme.

### KANUN KAYNAKLARI:
{baglam}

### SORU:
{soru}

### HUKUKİ CEVAP:
"""

    inputs = tokenizer([prompt], return_tensors = "pt").to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens = 300,
            temperature = 0.0,
            do_sample = False,
            repetition_penalty = 1.0,
            use_cache = False
        )

    res = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return res.split("### HUKUKİ CEVAP:")[-1].strip()

# TEST
print("🔥 KRİTİK TEST (Kasten Yaralama):")
print(hukuk_asistani_sor("Kasten birini yaralamanın cezası nedir?"))
```
